# TÊMPERA SIMULADA
> Um algoritmo de busca local, que busca um ótimo global evitando de ficar preso em máximos locais



```python
import random
import math

# Função objetivo
def valor(x):
    return -abs(x - 7)

# Escalonamento exponencial
def escalonamento(t, T0=10, alpha=0.95):
    return T0 * (alpha ** t)

def tempera_simulada():
    atual = random.randint(5, 8)
    print(f"Início em: {atual}")

    for t in range(100):
        T = escalonamento(t)
        print("tt", t)
        print("T", T)


        if T < 0.001:
            break

        # escolhe vizinho aleatório (+1 ou -1)
        proximo = atual + random.choice([-1, 1])
        print("proximo", proximo)


        # manter dentro do intervalo
        if proximo < 0 or proximo > 20:
            continue

        deltaE = valor(proximo) - valor(atual)
        print("delta E:", deltaE)

        # se melhorou, aceita
        if deltaE > 0:
            atual = proximo
        else:
            # aceita com probabilidade
            prob = math.exp(deltaE / T)
            if random.random() < prob:
                atual = proximo

    return atual

resultado = tempera_simulada()
print("Resultado final:", resultado)
```
