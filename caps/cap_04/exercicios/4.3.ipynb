## Impelmentação de subida de encosta:
> Algoritmo utilizado (caps/cap_04/alg_busca_subida_encosta)

- Para esse algortitmo precisamos da seguinte estrutura:

problema = {
    'estado_inicial': estado_inicial(len(cidades)),
    'sucessores': sucessores,
    'valor': lambda rota: valor(rota, cidades)
}


```python
def distancia_total(rota, cidades):
    total = 0
    for i in range(len(rota)):
        '''
            Explicação c1:
                Com base nas cidades retornadas pelo gerador de instancias
                e com a rota formada, assim pegando o indice da rota com base na 
                iteração atual e esse indice da rota representa o inidce da cidade o
                qual ele vai pegar
        '''
        c1 = cidades[rota[i]]
        
        '''
            Explicação c2:
                Ele sempre pega a cidade depois da de c1 e caso passe do comprimento de indices
                da rota ele volta para o começo para fechar o ciclo TSP
        '''
        c2 = cidades[rota[(i+1) % len(rota)]]

        '''
            .distancia é o método do arquivo importado em: gerador_instancia_tsp
        '''
        total += c1.distancia(c2)
    return total


def valor(rota, cidades):
    return -distancia_total(rota, cidades)

def sucessores(rota):

    '''
        Dentro dessa função ele gera todos os sucessores
        com base no estado atual
    '''
    vizinhos = []

    n = len(rota)

    for i in range(n):
        for j in range(i+1, n):
            nova = rota.copy()
            nova[i], nova[j] = nova[j], nova[i]
            vizinhos.append(nova)
        
    
    return vizinhos

def estado_inicial(n):
    '''
        Com base na quantidade de ciades geradas pelo gerador de instancias
        ele gera uma lista embaralhada de forma aleatoria
    '''
    rota = list(range(n))
    random.shuffle(rota)

    return rota

def subida_encosta(problema):
    '''
        Chama a função estado inicial, passsando o len de cidades
    '''
    corrente = problema['estado_inicial']

    while True:
        '''
            Chama a função sucessores, passando o estado atual como parametro
        '''
        vizinhos = problema['sucessores'](corrente)

        if not vizinhos:
            return corrente
        
        '''
            Chama a função lambda, ou seja o valor maximo em vizinhos que é passado posteriormente 
            para função valor
        '''
        vizinho = max(vizinhos, key=problema['valor'])

        if problema['valor'](vizinho) <= problema['valor'](corrente):
            return corrente
        
        corrente = vizinho

cidades = gerar_tsp(5)



problema = {
    'estado_inicial': estado_inicial(len(cidades)),
    'sucessores': sucessores,
    'valor': lambda rota: valor(rota, cidades)
}


melhor_rota = subida_encosta(problema)
print(melhor_rota)
```
