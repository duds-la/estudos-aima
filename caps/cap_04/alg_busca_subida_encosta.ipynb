
# BUSCA DE SUBIDA DE ENCONSTA
> Um laço de repetição que avalia os seus vizinhos e escolhe o de amior valor afim de encontrar um "pico", ele não guarda uma árvore de busca;




```python
def subida_encosta(problema):

    corrente = problema['estado_inicial']

    while True:
        vizinhos = problema['sucessores'](corrente)## chama a função para gerar sucessores

        if not vizinhos:
            return corrente
        
        vizinho = max(vizinhos, key=problema['valor'])

        if problema['valor'](vizinho) <= problema['valor'](corrente):
            return corrente
        
        corrente = vizinho
```
