## Busca Bidirecional
> Conceito: A busca bidirecional faz duas buscas simultâneas: Apartir do nó raiz e do nó objetivo, quando ambas as buscas se encontram ela para.



---

```python
import networkx as nx
from grafo_romenia import criar_grafo_romenia

G = criar_grafo_romenia()
```

```python
no_raiz = "Arad"
no_objetivo = "Bucharest"

busca_bidirecional = nx.bidirectional_shortest_path(G, no_raiz, no_objetivo)


print(list(busca_bidirecional))
```
