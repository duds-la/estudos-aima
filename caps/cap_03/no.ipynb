## NÓ:

#### Conceito:
> Nó nada mais é que um vértice, que também é uma entidade de conhecimento individual, ou seja o nó é a representação de alguns dados/valores

- Para criarmos o nó podemos ter duas abordagens:
    1. Criar uma classe para instanciar nó a nó
    2. Usar uma biblioteca como `networkx`



### Classe:

```python
class No:
    def __init__(self, id, valor=None):
        self.id = id       # identificador do nó
        self.valor = valor # dado adicional
        self.vizinhos = [] # lista de nós conectados

    def adicionar_vizinho(self, outro_no):
        self.vizinhos.append(outro_no)

# Criando nós
n1 = No("A")
n2 = No("B")
n3 = No("C")

# Conectando nós
n1.adicionar_vizinho(n2)
n1.adicionar_vizinho(n3)
n2.adicionar_vizinho(n1)
n3.adicionar_vizinho(n1)

# Acessando vizinhos
print([v.id for v in n1.vizinhos])  
```

### Exibição Gráfica da classe Nó:

```python
import matplotlib.pyplot as plt

# Classe No
class No:
    def __init__(self, id, valor=None):
        self.id = id
        self.valor = valor
        self.vizinhos = []

    def adicionar_vizinho(self, outro_no):
        self.vizinhos.append(outro_no)

# Criando nós
n1 = No("A")
n2 = No("B")
n3 = No("C")

# Conectando nós
n1.adicionar_vizinho(n2)
n1.adicionar_vizinho(n3)

# Lista de nós
nodes = [n1, n2, n3]

# Definindo posições (manual ou simples)
pos = {
    "A": (1,1),
    "B": (0, -1),
    "C": (2, -1)
}

# Desenhar nós
for node in nodes:
    x, y = pos[node.id]
    plt.scatter(x, y, s=1000, color='lightblue')
    plt.text(x, y, node.id, fontsize=14, ha='center', va='center')

# Desenhar arestas
for node in nodes:
    x1, y1 = pos[node.id]
    for viz in node.vizinhos:
        x2, y2 = pos[viz.id]
        plt.plot([x1, x2], [y1, y2], color='gray', linewidth=1)

plt.axis('off')
plt.show()
```

## Implementação utilizando a biblioteca `networkx`:

```python
import networkx as nx
import matplotlib.pyplot as plt


# Criar grafo
G = nx.Graph()

# Adicionar nós
G.add_node("A", valor=10)
G.add_node("B", valor=20)
G.add_node("C", valor=30)

# Adicionar arestas (conexões)
G.add_edge("A", "B")
G.add_edge("A", "C")

# Mostrar vizinhos de A
print(list(G.neighbors("A")))  # ['B', 'C']

# Mostrar todos os nós e atributos
print(G.nodes(data=True))
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Criar grafo
G = nx.Graph()

# Adicionar nós
G.add_node("A", valor=10)
G.add_node("B", valor=20)
G.add_node("C", valor=30)

# Adicionar arestas (conexões)
G.add_edge("A", "B")
G.add_edge("A", "C")

# Mostrar vizinhos de A
print(list(G.neighbors("A")))  # ['B', 'C']

# Mostrar todos os nós e atributos
print(G.nodes(data=True))

# -----------------------------
# Exibir grafo visualmente
# -----------------------------
pos = nx.spring_layout(G)  # calcula posições automáticas para os nós

# Desenhar nós
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=1500)

# Desenhar arestas
nx.draw_networkx_edges(G, pos, width=2)

# Desenhar rótulos
nx.draw_networkx_labels(G, pos, font_size=14, font_color='black')

# Se quiser, pode mostrar os atributos dos nós
node_labels = nx.get_node_attributes(G, 'valor')
nx.draw_networkx_labels(G, pos, labels=node_labels, font_color='red', verticalalignment='bottom')

plt.axis('off')  # desliga eixos
plt.show()
```
