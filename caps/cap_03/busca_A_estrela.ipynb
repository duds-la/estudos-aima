## A* (Busca A estrela)
> Conceito: Ela avalia os nós atráves da combinação de:
> - g(n) -> Custo para alcançar o nó;
> - h(n) -> Custo para ir do nó ao nó Objetivo;
> 
> **f(n) = g(n) + h(n)**
> 
> Uma vez que g(n) dá o custo do caminho desde o nó inicial até o nó n e h(n) é o custo estimado do
> caminho de menor custo de n até o objetivo, teremos
> 
> **f (n) = custo estimado da solução de menor custo através de n**



---

```python
import networkx as nx
from grafo_romenia import criar_grafo_romenia
import math

G = criar_grafo_romenia()

def heuristica(n1, n2):
    x1, y1 = G.nodes[n1]["pos"]
    x2, y2 = G.nodes[n2]["pos"]
    return math.sqrt((x1 - x2)**2 + (y1 - y2)**2)

inicio = "Arad"
objetivo = "Bucharest"

caminho = nx.astar_path(
    G,
    source=inicio,
    target=objetivo,
    heuristic=heuristica,
    weight="peso"
)

custo = nx.astar_path_length(
    G,
    source=inicio,
    target=objetivo,
    heuristic=heuristica,
    weight="peso"
)

print("Caminho encontrado:", caminho)
print("Custo total:", custo)
```
