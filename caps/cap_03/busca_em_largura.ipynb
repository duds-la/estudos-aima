## BFS (Busca em largura)
> Conceito: O nó raiz é expandido em seguida os sucessores desse nó e assim por diante. Em geral todos os nós de uma dada profundidade são expandidos antés que se passe para os nós do próximo nível;



### Criado arquivo python **grafo_romenia** para ajudar e evitar repetição de código

```python
import networkx as nx
from grafo_romenia import criar_grafo_romenia

G = criar_grafo_romenia()

print(G.nodes(data=True))
print(G.edges(data=True))
```

```python
no_raiz = 'Arad'
arvore = nx.bfs_tree(G, source=no_raiz)

print("Nós em árvore BFS:", list(arvore.nodes()))
print("Arestas em árvore BFS:", list(arvore.edges()))
```
